### 1. Error y control de temperatura

El error de control es $e[k]=r[k]-y[k]$. **Pregunta:** ¿qué valor toma si la referencia es $40^\circ C$ y el sensor marca $34^\circ C$?

```python
referencia=40.0
medida=34.0
error=referencia-medida
print(f"Error = {error:.1f} °C")
```

### 2. Respuesta al escalón de una planta de primer orden

Modelo **didáctico** $T(t)=T_0+K_p\Delta u(1-e^{-(t-t_0-\theta)/\tau})$ para $t\ge t_0+\theta$. Ajuste los parámetros y explique los cambios.

```python
t=np.arange(0,501,2)
T0,Kp,tau,theta,t0,du=23,0.55,125,6,30,40
T=T0+Kp*du*(1-np.exp(-np.maximum(t-t0-theta,0)/tau))
plt.plot(t,T,label="T1 calculada")
plt.xlabel("Tiempo (s)");plt.ylabel("Temperatura (°C)");plt.grid();plt.legend();plt.show()
```
