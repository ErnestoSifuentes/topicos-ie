Calcule la frecuencia efectiva a partir de las marcas de tiempo y examine las variaciones del período de muestreo.

```python
import csv, io, numpy as np, matplotlib.pyplot as plt
# Copie aquí sus filas CSV o cargue el archivo en Colab.
texto = """t_ms,adc,voltage
0,1900,1.5311
100,2000,1.6117
200,2100,1.6923
300,2200,1.7729
400,2300,1.8535
"""
filas=list(csv.DictReader(io.StringIO(texto)))
t=np.array([int(r['t_ms']) for r in filas])/1000
v=np.array([float(r['voltage']) for r in filas])
assert len(t)>1 and np.all(np.diff(t)>0)
print('Registros:',len(t),'| Período medio (ms):',round(float(np.mean(np.diff(t)))*1000,2))
plt.figure(figsize=(8,3));plt.plot(t,v,'o-');plt.xlabel('Tiempo (s)');plt.ylabel('Voltaje estimado (V)');plt.grid(alpha=.3);plt.show()
```
